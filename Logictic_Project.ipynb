```python
import pandas as pd 
import numpy as np
```

```python
orders = pd.read_excel("Company ShopX - Order Report.xlsx")
weight = pd.read_excel("Company ShopX - Product Weight.xlsx")
zone = pd.read_excel("Company ShopX - Warehouse&Customer Pin Code and Zone details.xlsx")
invoice = pd.read_excel("Courier Company - Invoice.xlsx")
rates = pd.read_excel("Courier Company - Rates.xlsx")
```

```python
print(orders.columns)
print(weight.columns)
print(zone.columns)
print(invoice.columns)
print(rates.columns)
```

```python

print(weight.head())
```

```python
df = pd.merge(orders, weight, on="Product Code")
df
```

```python
df["Total Weight (g)"] = df["Product Weight (g)"] *df["Units Ordered"]
```

```python
df
```

```python
Order_Weight = df.groupby("Order ID")['Total Weight (g)'].sum().reset_index()
```

```python
Order_Weight
```

```python
final_df = pd.merge(Order_Weight, invoice, on="Order ID")
```

```python
final_df
```

```python
# To conver gram into KG
final_df["Total Weight (KG)"] = final_df["Total Weight (g)"] / 1000
final_df["Courier Weight (KG)"] = final_df["Chargeable Weight"] 
```

```python
final_df
```

```python
# Makes Slab like 2 KG so there are 4 slabs of 0.5KG
final_df["Slab Count"] = np.ceil(final_df["Total Weight (KG)"] / 0.5)
```

```python
final_df
```

```python
rates.columns = rates.columns.str.strip()
```

```python
zone.head()
```

```python
rates.head()
```

```python
def calculate_charge(row):
    slabs = row["Slab Count"]
    zone = row["Delivery Zone"].lower()   # a, b, c...
    
    # Dynamic column names
    fwd_fixed_col = f"fwd_{zone}_fixed"
    fwd_add_col = f"fwd_{zone}_additional"
    
    rto_fixed_col = f"rto_{zone}_fixed"
    rto_add_col = f"rto_{zone}_additional"
    
    # Forward charge
    charge = rates.iloc[0][fwd_fixed_col]
    
    if slabs > 1:
        charge += (slabs - 1) * rates.iloc[0][fwd_add_col]
    
    # RTO condition
    if row["Freight Type"] == "Forward and RTO charges":
        charge += rates.iloc[0][rto_fixed_col]
        
        if slabs > 1:
            charge += (slabs - 1) * rates.iloc[0][rto_add_col]
    
    return charge
```

```python
final_df["Expected Charge (Rs.)"] = final_df.apply(calculate_charge, axis=1)
```

```python
final_df
```

```python
final_df["Difference (Rs.)"] = final_df["Total Amount (Rs.)"] - final_df["Expected Charge (Rs.)"]
final_df
```

```python
final_df["Weight slab as per ShopX (KG)"] = np.ceil(final_df["Total Weight (KG)"] / 0.5) * 0.5
final_df["Weight slab charged by Courier (KG)"] = np.ceil(final_df["Courier Weight (KG)"] / 0.5) * 0.5
final_df
```

```python
def status(x):
    if x == 0:
        return "Correct"
    elif x > 0:
        return "Overcharged"
    else:
        return "Undercharged"

final_df["Status"] = final_df["Difference (Rs.)"].apply(status)
```

```python
final_df
```

# Summary Table 

```python
correct = final_df[final_df["Status"]=="Correct"]
over = final_df[final_df["Status"]=="Overcharged"]
under = final_df[final_df["Status"]=="Undercharged"]
```

```python
summary = pd.DataFrame({
    'Type': ['Correct','Overcharged','Undercharged'],
    'Count': [len(correct), len(over), len(under)],
    'Amount': [correct['Expected Charge (Rs.)'].sum(),
               over['Difference (Rs.)'].sum(),
               under['Difference (Rs.)'].sum() * -1]
})
```

```python
summary
```

```python
output = pd.DataFrame()

output["Order ID"] = final_df["Order ID"]
output["AWB Number"] = final_df["AWB Code (Airway Bill Number) "].astype(str)

output["Total weight as per ShopX (KG)"] = final_df["Total Weight (KG)"]
output["Weight slab as per ShopX (KG)"] = final_df["Weight slab as per ShopX (KG)"]

output["Total weight as per Courier Company (KG)"] = final_df["Courier Weight (KG)"]
output["Weight slab charged by Courier Company (KG)"] = final_df["Weight slab charged by Courier (KG)"]

output["Delivery Zone as per ShopX"] = final_df["Delivery Zone"]
output["Delivery Zone charged by Courier Company"] = final_df["Delivery Zone"]

output["Expected Charge as per ShopX (Rs.)"] = final_df["Expected Charge (Rs.)"]
output["Charges Billed by Courier Company (Rs.)"] = final_df["Total Amount (Rs.)"]

output["Difference Between Expected Charges and Billed Charges (Rs.)"] = final_df["Difference (Rs.)"].round(2)
```

```python
output
```

```python
with pd.ExcelWriter("Logistic_Final_Report_20.xlsx") as writer:
    output.to_excel(writer, sheet_name="output", index=False)
    summary.to_excel(writer, sheet_name="Summary", index=False)
```
